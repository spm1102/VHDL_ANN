```python
from model import *
from binarization import FixedThreshold, OtsuThreshold, SauvolaThreshold

model1 = model1()
model2 = model2()
model3 = model3()

models = (model1, model2, model3)

binarization_cfgs = {
    "fixed":   FixedThreshold(threshold=0.5),
    "otsu":    OtsuThreshold(),
    "sauvola": SauvolaThreshold(window_size=15, k=0.2, R=0.5),
}
```

```python
import os
from helpers import train_loop, save_model_params


for bin_name, bin_transform in binarization_cfgs.items():
    print(f" Training with {bin_name.upper()} binarization")

    train_dataset, test_dataset = take_dataset(
        download=True,
        threshold_transform=bin_transform
    )
    # plot_data(train_dataset)
    train_dataloader, test_dataloader = take_dataloader(
        train_data=train_dataset,
        test_data=test_dataset,
        batch_size=BATCH_SIZE
    )

    outdir = f"weights_{bin_name}"
    os.makedirs(outdir, exist_ok=True)

    for model in models:
        print(f"\nModel: {model.__class__.__name__} | Binarization: {bin_name}")

        train_loop(
            model=model,
            epochs=EPOCHS,
            train_dataloader=train_dataloader,
            test_dataloader=test_dataloader,
            device=device
        )

        best_path = f"{model.__class__.__name__}_best.pth"
        model.load_state_dict(torch.load(best_path, map_location=device))

        save_model_params(model, outdir=outdir)

# for model in models:
#     train_loop(
#         model=model,
#         epochs=EPOCHS,
#         train_dataloader=train_dataloader,
#         test_dataloader=test_dataloader,
#         device=device
#     )
#     best_path = f"{model.__class__.__name__}_best.pth"
#     model.load_state_dict(torch.load(best_path))
#     save_model_params(model, outdir=weights)
```
